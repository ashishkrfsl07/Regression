# Q1.: What is Simple Linear Regression?


Simple Linear Regression (SLR) is a statistical method used to model the relationship between two continuous variables: one independent variable (or predictor, denoted as x) and one dependent variable (or response, denoted as y).

The goal of SLR is to find a straight line that best fits the observed data, allowing you to:

    Summarize the strength and direction of the linear relationship between the variables.

Predict the value of the dependent variable (y) based on the value of the independent variable (x).

It is called "simple" because it involves only one predictor variable. In contrast, "multiple linear regression" uses two or more predictor variables.

# Q2.:  What are the key assumptions of Simple Linear Regression?


The validity and reliability of the results from Simple Linear Regression (SLR) depend on several key assumptions about the data and the error term (ϵ) being met. When these assumptions are violated, the parameter estimates (β0​ and β1​) may be inefficient, and the associated hypothesis tests (p-values) and confidence intervals may be inaccurate.

The four primary assumptions are often summarized by the acronym LINE:

Key Assumptions of Simple Linear Regression

1. Linearity: The L in LINE

    Assumption: The relationship between the independent variable (x) and the mean of the dependent variable (y) must be linear.
2. Independence of Errors: The I in LINE

    Assumption: The error terms (ϵ) must be independent of each other.

    What it means: The residual for one observation is not correlated with the residual for any other observation. The error terms are not systematically related.

    Violation Indicator: This is most often a concern with time-series data, where the error at one point in time might be correlated with the error at the next point in time (autocorrelation). A plot of residuals versus time (or observation order) shows a clear pattern (e.g., cyclical or consistently increasing/decreasing).
3. Normality of Errors: The N in LINE

    Assumption: The error terms (ϵ) are normally distributed (with a mean of zero).
    What it means: The distribution of the residuals should approximate the bell-shaped curve.

    Violation Indicator: This is checked using a histogram of the residuals or a Normal Q-Q Plot. Significant skewness or heavy tails on the plot suggest a violation. While not strictly necessary for obtaining the coefficient estimates, this assumption is essential for conducting valid hypothesis tests and constructing confidence intervals.
4. Equal Variance (Homoscedasticity): The E in LINE

    Assumption: The variance of the error terms (ϵ) must be constant for all values of the independent variable (x). This is known as homoscedasticity (meaning "same scatter")
    What it means: The spread of the residuals should be uniform across the entire range of the predicted values.

    Violation Indicator: When the spread of the residuals increases or decreases as the predicted value increases (a funnel or cone shape), the assumption is violated. This non-constant variance is called heteroscedasticity.

# Q3.:  What is heteroscedasticity, and why is it important to address in regression models?

Heteroscedasticity (pronounced: het-uh-roh-skuh-das-tiss-i-tee) is the term used in regression analysis to describe a condition where the variance (or spread) of the error term (residuals) is not constant across all levels of the independent variable(s).

    Heteroscedasticity means unequal scatter (variance is not constant).

    Its opposite, which is a key assumption of Ordinary Least Squares (OLS) regression, is Homoscedasticity (equal scatter/constant variance).

Visualizing the Problem

When you plot the residuals against the predicted values (y^​) or the independent variable (x), heteroscedasticity is typically indicated by a distinctive pattern:

    Fan or Cone Shape: The spread of the residuals either systematically widens (most common) or narrows as the predicted value increases.

    A homoscedastic plot would show the residuals scattered randomly and uniformly across the entire range, resembling a horizontal band.

A common real-world example is a regression modeling family income and expenditure on non-essentials. Lower-income families have little variability in non-essential spending, but higher-income families show a much wider range of spending habits, leading to larger residuals at higher income levels.

Why is it Important to Address?

Heteroscedasticity is important to address because, while it does not bias the regression coefficient estimates (β0​,β1​), it significantly impacts the reliability of the statistical inference derived from the model.

How to Address Heteroscedasticity

Common solutions to address heteroscedasticity include:

    Heteroscedasticity-Consistent Standard Errors (HCSE) / Robust Standard Errors: This is the most common approach. It adjusts the standard error calculations (e.g., using the Huber-White method) without changing the estimated coefficients. This fixes the inference problem, especially with large samples.

Weighted Least Squares (WLS): This method assigns less weight to observations associated with a larger error variance and more weight to observations with smaller error variance, effectively transforming the model to satisfy the constant variance assumption.

Variable Transformation: Applying a transformation, such as taking the logarithm or square root of the dependent variable (y), can often stabilize the variance.

# Q4.: What is Multiple Linear Regression?

Multiple Linear Regression (MLR) is an extension of Simple Linear Regression (SLR) that is used to model the linear relationship between one continuous dependent variable (the outcome or response, denoted as Y) and two or more independent variables (the predictors or explanatory variables, denoted as X1​,X2​,…,Xk​).

In essence, MLR allows analysts to:

    Determine the overall strength of the linear relationship between the dependent variable and the entire set of independent variables.

    Estimate how much each individual independent variable contributes to the prediction of the dependent variable, while controlling for the effects of all other independent variables in the model. This is the key advantage over SLR.

    Predict the value of the dependent variable for a given set of values for the independent variables.

The MLR Model Formula

The mathematical equation for the Multiple Linear Regression model is:

Y^=β0​+β1​X1​+β2​X2​+…+βk​Xk​+ϵ

Where:

    Y^ is the predicted value of the dependent variable.
    X1​,X2​,…,Xk​ are the independent variables (the predictors).
    β0​ is the Y-intercept (the expected value of Y when all X variables are zero).
    β1​,β2​,…,βk​ are the partial regression coefficients (or partial slopes). βi​ represents the change in Y for a one-unit increase in Xi​, assuming all other independent variables (X values) are held constant.
    ϵ is the error term or residual, representing the unexplained variation.

## Q5.:  What is polynomial regression, and how does it differ from linear regression?

Polynomial Regression is a form of Linear Regression in which the relationship between the independent variable (X) and the dependent variable (Y) is modeled as an nth degree polynomial. Unlike simple linear regression, which models the relationship using a straight line (Y^=β0​+β1​X), polynomial regression introduces squared, cubed, or higher-order terms of the independent variable, such as Y^=β0​+β1​X+β2​X2+β3​X3+…+βn​Xn. The key difference is that linear regression strictly models a straight-line relationship, while polynomial regression models a curved relationship, making it suitable for datasets where the relationship is not simple and monotonic. Crucially, although it models a curved relationship, polynomial regression is still considered "linear" in the sense of the coefficients (β), which are estimated linearly using the same Ordinary Least Squares (OLS) method, hence it remains within the general linear model framework.

# Q6.: Implement a Python program to fit a Simple Linear Regression model to the following sample data:
# ● X = [1, 2, 3, 4, 5]
# ● Y = [2.1, 4.3, 6.1, 7.9, 10.2]
# Plot the regression line over the data points.


```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Sample Data
X = np.array([1, 2, 3, 4, 5])
Y = np.array([2.1, 4.3, 6.1, 7.9, 10.2])
n = len(X)

# 2. Calculate Regression Coefficients (OLS Method)

# Calculate the mean of X and Y
mean_x = np.mean(X)
mean_y = np.mean(Y)

# Calculate the slope (beta_1)
# Formula: beta_1 = sum((Xi - mean_x) * (Yi - mean_y)) / sum((Xi - mean_x)^2)
numerator = np.sum((X - mean_x) * (Y - mean_y))
denominator = np.sum((X - mean_x)**2)
beta_1 = numerator / denominator

# Calculate the intercept (beta_0)
# Formula: beta_0 = mean_y - beta_1 * mean_x
beta_0 = mean_y - (beta_1 * mean_x)

# 3. Create the Regression Model
# The equation is: Y_hat = beta_0 + beta_1 * X
Y_hat = beta_0 + beta_1 * X

# 4. Output the Results
print(f"Independent Variable (X): {X}")
print(f"Dependent Variable (Y): {Y}")
print("-" * 40)
print(f"Intercept (β₀): {beta_0:.4f}")
print(f"Slope (β₁): {beta_1:.4f}")
print("-" * 40)
print(f"Regression Equation: Y_hat = {beta_0:.4f} + {beta_1:.4f} * X")
print("-" * 40)

# 5. Plot the Results
plt.figure(figsize=(8, 6))

# Plot the actual data points
plt.scatter(X, Y, color='blue', label='Actual Data Points')

# Plot the regression line
plt.plot(X, Y_hat, color='red', label='Regression Line')

# Add labels and title
plt.xlabel('Independent Variable (X)')
plt.ylabel('Dependent Variable (Y)')
plt.title('Simple Linear Regression Fit')
plt.legend()
plt.grid(True)
plt.show()
```

# Q7.: Fit a Multiple Linear Regression model on this sample data:
# ● Area = [1200, 1500, 1800, 2000]
# ● Rooms = [2, 3, 3, 4]
# ● Price = [250000, 300000, 320000, 370000]
# Check for multicollinearity using VIF and report the results

```python
import pandas as pd
import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Sample data
data = {
    'Area': [1200, 1500, 1800, 2000],
    'Rooms': [2, 3, 3, 4],
    'Price': [250000, 300000, 320000, 370000]
}
df = pd.DataFrame(data)

# Define independent variables (X) and dependent variable (y)
X = df[['Area', 'Rooms']]
y = df['Price']

# Add a constant to the independent variables for the OLS model
X = sm.add_constant(X)

# Fit the Multiple Linear Regression model (OLS)
model = sm.OLS(y, X).fit()

# Calculate VIF for multicollinearity check
# We iterate over the columns of the design matrix X (excluding the constant at index 0)
vif_data = pd.DataFrame()
vif_data["Variable"] = X.columns
vif_data["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]

# Print the model summary and the VIF results
print("--- Multiple Linear Regression Model Summary ---")
print(model.summary())
print("\n--- Variance Inflation Factor (VIF) Results ---")
print(vif_data)
```

# Q8.: Implement polynomial regression on the following data:
# ● X = [1, 2, 3, 4, 5]
# ● Y = [2.2, 4.8, 7.5, 11.2, 14.7]
# Fit a 2nd-degree polynomial and plot the resulting curve.


```python
import numpy as np
import matplotlib.pyplot as plt

# Data
X = np.array([1, 2, 3, 4, 5])
Y = np.array([2.2, 4.8, 7.5, 11.2, 14.7])

# Fit a 2nd-degree polynomial (Y = aX^2 + bX + c)
# The output is [a, b, c]
coefficients = np.polyfit(X, Y, 2)
# Create a polynomial function from the coefficients
poly_function = np.poly1d(coefficients)

# Generate points for the fitted curve
# Create a smooth range of x values
X_fit = np.linspace(X.min(), X.max(), 100)
# Calculate the predicted y values
Y_fit = poly_function(X_fit)

# Plotting
plt.figure(figsize=(8, 5))
# Plot original data points
plt.scatter(X, Y, color='red', label='Original Data Points')
# Plot the fitted polynomial curve
plt.plot(X_fit, Y_fit, color='blue', label='2nd-Degree Polynomial Fit')

plt.title('2nd-Degree Polynomial Regression Fit')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.savefig('polynomial_regression_plot.png')
plt.close()

print(f"Polynomial Coefficients (a, b, c): {coefficients}")
```

# Q9.: Create a residuals plot for a regression model trained on this data:
# ● X = [10, 20, 30, 40, 50]
# ● Y = [15, 35, 40, 50, 65]
# Assess heteroscedasticity by examining the spread of residuals.


```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt

# 1. Prepare the data
X = [10, 20, 30, 40, 50]
Y = [15, 35, 40, 50, 65]

df = pd.DataFrame({'X': X, 'Y': Y})

# 2. Train the model: Add a constant to the independent variable for the intercept
X_with_const = sm.add_constant(df['X'])
model = sm.OLS(df['Y'], X_with_const).fit()

# 3. Calculate residuals
y_pred = model.predict(X_with_const)
residuals = model.resid

# 4. Create the plot
plt.figure(figsize=(8, 6))
# Scatter plot of residuals vs. predicted values
plt.scatter(y_pred, residuals, color='blue')
# Add a horizontal line at 0 for reference
plt.hlines(0, y_pred.min(), y_pred.max(), color='red', linestyle='--')
plt.title('Residuals Plot')
plt.xlabel('Predicted Values ($\hat{Y}$)')
plt.ylabel('Residuals ($Y - \hat{Y}$)')
plt.grid(True)
plt.savefig('residuals_plot.png')
plt.close()

# Prepare data for display and assessment
residuals_df = pd.DataFrame({'Predicted_Y': y_pred, 'Residuals': residuals})
print(residuals_df)
```

# Q10.:  Imagine you are a data scientist working for a real estate company. You need to predict house prices using features like area, number of rooms, and location. However, you detect heteroscedasticity and multicollinearity in your regression model. Explain the steps you would take to address these issues and ensure a robust model.


That's a common challenge in real estate data! Dealing with heteroscedasticity and multicollinearity is crucial for building a reliable and robust house price prediction model.

Here are the steps I would take to address these issues:

🏠 Addressing Heteroscedasticity

Heteroscedasticity occurs when the variance of the errors (residuals) in your model is not constant across all levels of the independent variables. This violates a core assumption of Ordinary Least Squares (OLS) regression, leading to inefficient and biased standard errors, making hypothesis tests (like t-tests) unreliable.

    1. Transform the Dependent Variable:

        Apply a transformation, most commonly the logarithmic transformation (e.g., ln(Price)), to the house price. This often stabilizes the variance and can also help normalize a skewed price distribution.

        The Box-Cox transformation is a more generalized approach that can determine the optimal transformation parameter.

    2. Use Weighted Least Squares (WLS):

        If the pattern of the heteroscedasticity can be modeled (e.g., variance is proportional to a feature like area), WLS assigns lower weights to observations with higher variance (larger errors) and higher weights to observations with lower variance. This provides more efficient estimates than OLS.

    3. Use Heteroscedasticity-Consistent Standard Errors (HCSE) or Robust Standard Errors:

        This is often the simplest and most robust approach. Instead of changing the model structure, HCSE (like the White estimator) adjusts the standard errors of the OLS coefficients to account for the non-constant variance. This doesn't change the coefficients themselves but makes the inference (p-values, confidence intervals) statistically valid.

🏘️ Addressing Multicollinearity

Multicollinearity occurs when two or more independent variables in a regression model are highly correlated. This makes it difficult for the model to estimate the unique effect of each variable, leading to high standard errors for the coefficients, unstable coefficient estimates (they can change drastically with minor data changes), and difficulty in interpreting the individual feature effects.

    1. Detect Multicollinearity:

        Calculate the Variance Inflation Factor (VIF) for each predictor. A common rule of thumb suggests that a VIF greater than 5 or 10 indicates a problematic level of multicollinearity.

    2. Remove or Combine Collinear Variables:

        If two variables (e.g., 'Area of the house' and 'Number of square feet on the lot') are highly correlated, remove one of them. Choose the one that is less theoretically relevant or less correlated with the dependent variable (Price).

        Create a composite feature: Combine the highly correlated features into a single, representative feature (e.g., instead of 'Number of full bathrooms' and 'Number of half bathrooms', use 'Total number of bathrooms').

    3. Use Penalized Regression Methods:

        Ridge Regression introduces a penalty term (λ) to the OLS cost function that shrinks the regression coefficients toward zero. This stabilizes the coefficient estimates and is very effective at managing multicollinearity, though it slightly biases the estimates.

        Lasso Regression is similar but can drive some coefficients exactly to zero, effectively performing feature selection and helping to create a more parsimonious model.

    4. Use Principal Component Regression (PCR):

        PCR transforms the set of correlated independent variables into a set of orthogonal (uncorrelated) variables called Principal Components. You then use a subset of these components (which capture most of the variance) as predictors in your regression model.

By systematically applying these techniques—starting with transformations or robust standard errors for heteroscedasticity and VIF analysis followed by variable selection or penalized regression for multicollinearity—I can ensure a robust, interpretable, and statistically valid predictive model for house prices.
